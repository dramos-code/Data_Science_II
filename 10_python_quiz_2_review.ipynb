## 4. `continue` versus `break`

### Expectation
- `continue` skips the **rest of the current iteration** and moves to the next.
- `break` exits the **nearest enclosing loop** entirely.

**Resource:** [Python break and continue](https://docs.python.org/3/tutorial/controlflow.html#break-and-continue-statements)

```python
print("Using continue:")
for number in range(1, 7):
    if number == 4:
        continue
    print(number)

print("Using break:")
for number in range(1, 7):
    if number == 4:
        break
    print(number)
```

### Try it

```python
for item in range(1, 6):
    if item == 2:
        continue
    print("A:", item)

for item in range(1, 6):
    if item == 2:
        break
    print("B:", item)
```

## Extra Practice: 2. While loops, stopping conditions, break and continue
`while` repeats while a condition is True. Update the counter or use a clearly reachable `break` so the loop ends. `continue` skips the rest of the **current** iteration; `break` exits the nearest loop.

```python
count = 0
while count < 4:
    count = count + 1  # update the value checked by while
    if count == 2:
        continue  # skip printing 2
    if count == 4:
        break  # exit the loop immediately
    print("Count:", count)

# Practice tracing a data cleaning total
scores = [62, 76, 91, 45]
passing = 0
for score in scores:
    if score < 70:
        continue
    passing = passing + 1
print("Passing:", passing)
```

## Extra Practice: 4. Functions and parameters
`def` creates a named reusable block. Parameters receive input values (arguments). A local variable inside a function belongs to that function.

```python
def describe_score(score):  # score is a parameter
    label = "Pass" if score >= 70 else "Retry"  # label is local
    print(label)

describe_score(82)  # 82 is an argument

def square(value):
    answer = value * value
    return answer  # sends the value back to the caller

result = square(5)
print("Square:", result)
```

## Extra Practice: 5. Exceptions: try and except
`try` attempts an operation. `except ValueError` handles a bad numeric conversion (for example, `int("hello")`) so the program can continue.

```python
raw_values = ["12", "oops", "7"]
for entry in raw_values:
    try:
        number = int(entry)  # converting nonnumeric text raises ValueError
        print("Valid:", number)
    except ValueError:
        print("Invalid entry:", entry)
```
