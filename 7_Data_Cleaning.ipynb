# 1. Clean Ages

For this activity, an age is valid if it is between `0` and `100`.

If the age is invalid, we will replace it with `None`.

`None` means we do not have a usable value.


````python
ages = [24, -5, 31, 150, 28, 42]

clean_ages = []

for age in ages:
    if age < 0:
        # .append() adds one new item to the end of a list.
        clean_ages.append(None)
    elif age > 100:
        clean_ages.append(None)
    else:
        clean_ages.append(age)

print("Original:", ages)
print("Cleaned: ", clean_ages)
````

## Check How Many Values We Have

Sometimes we want to count how many observations are in a list.


````python
# len(...) tells us how many items are inside a list.
print("Number of original ages:", len(ages))
print("Number of cleaned ages:", len(clean_ages))
````

A product quantity should not be negative.

`0` is allowed because an item can be out of stock.


````python
quantities = [12, 8, -3, 0, 25, -1]

clean_quantities = []

for quantity in quantities:
    if quantity < 0:
        clean_quantities.append(None)
    else:
        clean_quantities.append(quantity)

print(clean_quantities)
````

# 4. Standardize States

Sometimes data is not wrong — it is just written in different formats.

We want:

- `New York` → `NY`
- `New Jersey` → `NJ`
- `Pennsylvania` → `PA`
- `California` → `CA`

Already-correct abbreviations stay the same.


````python
states = ["NY", "New Jersey", "PA", "New York", "CA", "Pennsylvania"]

clean_states = []

for state in states:
    if state == "New York":
        clean_states.append("NY")
    elif state == "New Jersey":
        clean_states.append("NJ")
    elif state == "Pennsylvania":
        clean_states.append("PA")
    elif state == "California":
        clean_states.append("CA")
    else:
        clean_states.append(state)

print("Original:", states)
print("Cleaned: ", clean_states)
````

# 5. Clean Yes / No Responses

The only accepted responses are:

- `"Yes"`
- `"No"`

Anything else becomes `None`.


````python
responses = ["Yes", "No", "Maybe", "Yes", "Unknown", "No"]

clean_responses = []

for response in responses:
    if response == "Yes":
        clean_responses.append(response)
    elif response == "No":
        clean_responses.append(response)
    else:
        clean_responses.append(None)

print(clean_responses)
````

# 6. Whole Numbers Only

Sometimes a value is numerical but should still be a whole number.

For example, years of experience might be:

- `5` → valid
- `-2` → invalid
- `3.5` → invalid


````python
experience = [2, 5, -1, 7, 2.5, 10, 0]

clean_experience = []

for value in experience:
    if value < 0:
        clean_experience.append(None)
    # int(value) removes the decimal part so we can check whether a number was already whole.
    elif value != int(value):
        clean_experience.append(None)
    else:
        clean_experience.append(value)

print(clean_experience)
````

# 7. Keep Only the Valid Values

Sometimes we may want a new list containing only valid observations.

This is different from keeping `None` as a placeholder.


````python
scores = [88, 105, 74, -6, 82, 100]

valid_scores = []

for score in scores:
    if score >= 0 and score <= 100:
        valid_scores.append(score)

print(valid_scores)
````
