## 1. Loop Through Observations

Imagine these are temperature observations collected over several days.

A `for` loop lets us look at one observation at a time.


````python
temperatures = [72, 75, 68, 80, 77]

for temperature in temperatures:
    print(temperature)
````

### What to notice
The loop runs once for each value in the list.

## 2. Transform Every Observation

Data scientists often apply the **same calculation** to every value.

Example: Convert Celsius temperatures to Fahrenheit.


````python
celsius = [20, 25, 30, 18, 22]

for temperature in celsius:
    fahrenheit = temperature * 9 / 5 + 32
    print(fahrenheit)
````

## 3. Build a New List

Sometimes we want to keep the transformed values.

We can:
1. start with an empty list
2. loop through the original data
3. calculate a new value
4. use `.append()` to save it


````python
monthly_income = [3000, 4200, 3800, 5000]

yearly_income = []

for income in monthly_income:
    yearly_income.append(income * 12)

print("Monthly income:", monthly_income)
print("Yearly income:", yearly_income)
````

## 4. Calculate a Total

Another common loop pattern is an **accumulator**.

We start with:

```python
total = 0
```

Then add each observation to the total.


````python
daily_sales = [250, 310, 180, 425, 390]

total_sales = 0

for sale in daily_sales:
    total_sales = total_sales + sale

print("Total sales:", total_sales)
````

You may also see:

```python
total_sales += sale
```

It means the same thing as:

```python
total_sales = total_sales + sale
```

## 5. Calculate an Average Manually

A mean is:

**total ÷ number of observations**

We can use a loop to calculate the total, then use `len()` to count the observations.


````python
scores = [82, 91, 76, 88, 93]

total = 0

for score in scores:
    total += score

average = total / len(scores)

print("Total:", total)
print("Number of scores:", len(scores))
print("Average:", average)
````

### Why this matters
Later, pandas can calculate an average for us.

But this helps us understand what a mean is doing underneath.

## 7. Repeat the Same Analysis for Several Columns

This is where loops become especially useful in data science.

Without a loop, we might write:

```python
df["Age"].mean()
df["Income"].mean()
df["Score"].mean()
```

With a loop, we write the action once.


````python
data = pd.DataFrame({
    "Age": [22, 28, 35, 41],
    "Income": [40000, 52000, 61000, 75000],
    "Score": [78, 85, 91, 88]
})

columns = ["Age", "Income", "Score"]

for column in columns:
    print(column)
    print(data[column].mean())
````

### What to notice
The variable `column` changes each time through the loop.

The same `.mean()` calculation is repeated for:
- Age
- Income
- Score

# Challenge 2 — Make Several Graphs Automatically

A loop can repeat plotting code too.

Instead of writing three separate histogram commands, we can loop through the column names.


````python
import matplotlib.pyplot as plt

columns = ["Age", "Income", "Score"]

for column in columns:
    plt.hist(data[column])
    plt.title(column)
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.show()
````
