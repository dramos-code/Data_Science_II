## Dataset setup — run this cell first
The table contains **360 synthetic bike-share trips** across 60 days. Variables include date, station, bike type, rider type, trip distance, duration, weather, fare and satisfaction. The numbers were created for practice and must not be treated as real-world evidence.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme()
rng = np.random.default_rng(2026)
n = 360
calendar = pd.date_range("2026-09-01", periods=60)
day_index = rng.choice(np.arange(60), size=n)
ride_dates = calendar[day_index]
bike_type = rng.choice(["Standard", "E-Bike"], size=n, p=[0.58, 0.42])
rider_type = rng.choice(["Member", "Casual"], size=n, p=[0.64, 0.36])
station = rng.choice(["Downtown", "University", "Harbor", "Park"], size=n)
time_of_day = rng.choice(["Morning", "Afternoon", "Evening"], size=n,
                         p=[0.40, 0.38, 0.22])
weather_temp = 23 + 5 * np.sin(np.linspace(0, 4, 60)) + rng.normal(0, 2, 60)
weather_rain = rng.choice([0, 0, 0, 0.8, 3.5, 8.0], size=60)
distance = np.clip(rng.gamma(3.0, 1.35, size=n), 0.5, 14.0)
duration = np.clip(3 + distance * np.where(bike_type == "E-Bike", 3.6, 5.2)
                   + rng.normal(0, 3, n) + weather_rain[day_index] * 0.30,
                   4, 95)
fare = 1.8 + distance * np.where(bike_type == "E-Bike", 1.0, 0.75)
fare += np.where(rider_type == "Casual", 1.4, 0)
satisfaction = np.clip(4.9 - 0.025 * duration - 0.04 * weather_rain[day_index]
                       + rng.normal(0, 0.48, n), 1, 5)

df = pd.DataFrame({
    "Trip_ID": np.arange(1001, 1001 + n),
    "Date": ride_dates,
    "Station": station,
    "Bike_Type": bike_type,
    "Rider_Type": rider_type,
    "Time_of_Day": time_of_day,
    "Distance_km": distance.round(2),
    "Duration_min": duration.round(1),
    "Temperature_C": weather_temp[day_index].round(1),
    "Rain_mm": weather_rain[day_index],
    "Fare_USD": fare.round(2),
    "Satisfaction": satisfaction.round(1)
}).sort_values("Date").reset_index(drop=True)
```
